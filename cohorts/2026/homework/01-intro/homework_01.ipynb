```python
#Loading data
df = pd.read_csv("car_fuel_efficiency_2026.csv")

#Total number of records
total_rows = len(df)
print(total_rows)
```

```python
# Finding unique value for fuel type
df.fuel_type.nunique()
```

```python
# Missing values
df.isnull().sum()
```

```python
# Max Fuel efficiency
print(df.groupby('origin')['fuel_efficiency_mpg'].max().loc['Asia'])
```

```python
#initial median
initial_horsepower = df['horsepower'].median()
print(f"initial horsepower median: {initial_horsepower}")
```

```python
# Most frequent value

most_frequent_hp = df['horsepower'].mode()[0]
print(f"most frequent hp: {most_frequent_hp}")
```

```python
# Filling missing values

df['horsepower'] = df['horsepower'].fillna(most_frequent_hp)
print(f"Remaining missing values: {df['horsepower'].isnull().sum()}")
```

```python
#Recalculating the median value

new_median = df['horsepower'].median()
print(f"new median value: {new_median}")
```

```python
# 1. Filter cars from Asia
asia_cars = df[df['origin'] == 'Asia']

# 2. Select only vehicle_weight and model_year columns
asia_subset = asia_cars[['vehicle_weight', 'model_year']]

# 3. Select the first 7 values
first_seven = asia_subset.iloc[:7]

# 4. Get the underlying NumPy array X
X = first_seven.to_numpy()

# 5. Compute matrix-matrix multiplication (X^T * X)
XTX = X.T.dot(X)

# 6. Invert XTX
XTX_inv = np.linalg.inv(XTX)

# 7. Create the vector array y
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])

# 8. Compute the Normal Equation formula weights (w)
w = XTX_inv.dot(X.T).dot(y)

# 9. Get the sum of all elements in w
result_sum = np.sum(w)
print(f"Sum of elements: {result_sum:.4f}")
```
